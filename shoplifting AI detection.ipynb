```python
import matplotlib.pyplot as plt
import numpy as np

# Get the metrics object
metrics = model.metrics

# Extract class-wise data (access attributes, don't call them!)
precision = metrics.box.p       # list of precision values per class
recall = metrics.box.r          # list of recall values per class
map50 = metrics.box.all_ap[:, 0]  # AP@0.5 per class (column 0 in all_ap)
map = metrics.box.all_ap[:, -1]   # AP@[.5:.95] per class (last column)

# Number of classes
num_classes = len(precision)
class_names = model.names.values() if hasattr(model, 'names') else [f'class_{i}' for i in range(num_classes)]

# Bar plot setup
x = np.arange(num_classes)
width = 0.2

# Create the plot
plt.figure(figsize=(12, 6))
plt.bar(x - 1.5*width, precision, width, label='Precision')
plt.bar(x - 0.5*width, recall, width, label='Recall')
plt.bar(x + 0.5*width, map50, width, label='mAP@0.5')
plt.bar(x + 1.5*width, map, width, label='mAP@0.5:0.95')


plt.xticks(x, class_names, rotation=45)
plt.ylabel('Score')
plt.title('Class-wise YOLOv8 Evaluation Metrics')
plt.ylim(0, 1.05)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()

```
